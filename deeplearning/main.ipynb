```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load the dataset
x_train = np.load("A4_data-20240409/hiragana_train_imgs.npz")["arr_0"]
y_train = np.load("A4_data-20240409/hiragana_train_labels.npz")["arr_0"]
x_test = np.load("A4_data-20240409/hiragana_test_imgs.npz")["arr_0"]
y_test = np.load("A4_data-20240409/hiragana_test_labels.npz")["arr_0"]

# Convert labels to one-hot encoding
y_train = keras.utils.to_categorical(y_train)
y_test = keras.utils.to_categorical(y_test)

# Define input shape
input_shape = (28, 28)

# Number of runs
num_runs = 10

# Lists to store accuracies
tanh_accs = []
relu_accs = []

# Run experiments
for i in range(num_runs):
    # Define the model with tanh activation
    print(i,"run \n")
    inputs = keras.Input(shape=input_shape)
    x = layers.Flatten()(inputs)
    x = layers.Dense(100, activation='tanh',
                     kernel_initializer=tf.keras.initializers.RandomNormal(),
                     bias_initializer=tf.keras.initializers.RandomNormal())(x)
    outputs = layers.Dense(10, activation='softmax',
                           kernel_initializer=tf.keras.initializers.RandomNormal(),
                           bias_initializer=tf.keras.initializers.RandomNormal())(x)
    model_tanh = keras.Model(inputs=inputs, outputs=outputs)
    model_tanh.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    model_tanh.fit(x_train, y_train, epochs=100, verbose=0)
    _, tanh_acc = model_tanh.evaluate(x_test, y_test, verbose=0)
    tanh_accs.append(tanh_acc)

    # Define the model with ReLU activation
    inputs = keras.Input(shape=input_shape)
    x = layers.Flatten()(inputs)
    x = layers.Dense(100, activation='relu',
                     kernel_initializer=tf.keras.initializers.RandomNormal(),
                     bias_initializer=tf.keras.initializers.RandomNormal())(x)
    outputs = layers.Dense(10, activation='softmax',
                           kernel_initializer=tf.keras.initializers.RandomNormal(),
                           bias_initializer=tf.keras.initializers.RandomNormal())(x)
    model_relu = keras.Model(inputs=inputs, outputs=outputs)
    model_relu.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    model_relu.fit(x_train, y_train, epochs=100, verbose=0)
    _, relu_acc = model_relu.evaluate(x_test, y_test, verbose=0)
    relu_accs.append(relu_acc)

# Calculate mean accuracies
tanh_mean_acc = np.mean(tanh_accs)
relu_mean_acc = np.mean(relu_accs)

print("Mean Test Accuracy with tanh activation:", tanh_mean_acc)
print("Mean Test Accuracy with ReLU activation:", relu_mean_acc)
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Load the dataset
x_train = np.load("A4_data-20240409/hiragana_train_imgs.npz")["arr_0"]
y_train = np.load("A4_data-20240409/hiragana_train_labels.npz")["arr_0"]
x_test = np.load("A4_data-20240409/hiragana_test_imgs.npz")["arr_0"]
y_test = np.load("A4_data-20240409/hiragana_test_labels.npz")["arr_0"]


# Convert labels to one-hot encoding
y_train = keras.utils.to_categorical(y_train)
y_test = keras.utils.to_categorical(y_test)

# Define input shape
input_shape = (28, 28)

# Define different configurations to explore
configurations = [
    # (1, [10]),
    # (1, [100]),
    # (1, [1000]),
    # (2, [10, 100])
    # (2, [100, 100]),
    # (2, [1000, 1000]),
    # (3, [10, 10, 10]),
    # (3, [100, 100, 100]),
    (3, [1000, 1000, 1000])
]

# Dictionary to store results
results = {}

# Perform initial runs for each configuration
for num_layers, num_units_list in configurations:
    test_accs = []
    for _ in range(1):  # Only one initial run
        # Define the model
        inputs = keras.Input(shape=input_shape)
        x = layers.Flatten()(inputs)
        for num_units in num_units_list:
            x = layers.Dense(num_units, activation='relu',
                             kernel_initializer=tf.keras.initializers.RandomNormal(),
                             bias_initializer=tf.keras.initializers.RandomNormal())(x)
        outputs = layers.Dense(10, activation='softmax',
                               kernel_initializer=tf.keras.initializers.RandomNormal(),
                               bias_initializer=tf.keras.initializers.RandomNormal())(x)
        model = keras.Model(inputs=inputs, outputs=outputs)
        
        # Compile the model
        model.compile(keras.optimizers.Adam(learning_rate=1e-5), loss='categorical_crossentropy', metrics=['accuracy'])
        
        # Train the model
        model.fit(x_train, y_train, epochs=20, verbose=0)
        
        # Evaluate the model on test data
        _, test_acc = model.evaluate(x_test, y_test, verbose=0)
        test_accs.append(test_acc)
        
    # Store mean test accuracy for the configuration
    mean_test_acc = np.mean(test_accs)
    results[(num_layers, tuple(num_units_list))] = mean_test_acc

# Print results
for config, mean_test_acc in results.items():
    print(f"Configuration: {config}, Mean Test Accuracy: {mean_test_acc}")

# # Identify the best configuration
# best_config = max(results, key=results.get)
# print("Best Configuration:", best_config)
# Configuration: (1, (10,)), Mean Test Accuracy: 0.3750999867916107
# Configuration: (1, (100,)), Mean Test Accuracy: 0.7361000180244446
# Configuration: (1, (1000,)), Mean Test Accuracy: 0.804099977016449
# Configuration: (2, (10, 10)), Mean Test Accuracy: 0.6323000192642212
# Configuration: (2, (100, 100)), Mean Test Accuracy: 0.777899980545044
# Configuration: (2, (1000, 1000)), Mean Test Accuracy: 0.8285999894142151
# Configuration: (3, (10, 10, 10)), Mean Test Accuracy: 0.6060000061988831
# Configuration: (3, (100, 100, 100)), Mean Test Accuracy: 0.8183000087738037
# Configuration: (3, (1000, 1000, 1000)), Mean Test Accuracy: 0.8346999883651733
```
